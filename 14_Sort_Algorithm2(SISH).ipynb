```python
import matplotlib.pyplot as plt
import numpy as np
import time

# 선택 정렬 함수
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

# 사용자에게 데이터 개수 입력 받기
data_size = int(input("정렬할 데이터의 개수를 입력하세요: "))

# 1부터 1000 사이의 랜덤한 정수를 data_size 개 생성
data = np.random.randint(1, 1001, size=data_size)
initial_data = data.copy()

# 선택 정렬 실행
start = time.time()
sorted_data = selection_sort(data)
end = time.time()

# 초기 상태와 최종 상태를 배열로 출력
print("초기 데이터:", initial_data)
print("정렬된 데이터:", sorted_data)
print(f"{end - start:.5f} sec")

# 초기 상태와 최종 상태 비교 (그래프)
plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.bar(range(len(initial_data)), initial_data, color='blue')
plt.title("Initial State")
plt.subplot(1, 2, 2)
plt.bar(range(len(sorted_data)), sorted_data, color='green')
plt.title("Final State")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import time

# 삽입 정렬 함수
def insertion_sort(arr):
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

# 사용자에게 데이터 개수 입력 받기
data_size = int(input("정렬할 데이터의 개수를 입력하세요: "))

# 1부터 1000 사이의 랜덤한 정수를 data_size 개 생성
data = np.random.randint(1, 1001, size=data_size)
initial_data = data.copy()

# 삽입 정렬 실행
start = time.time()
sorted_data = insertion_sort(data)
end = time.time()

# 초기 상태와 최종 상태를 배열로 출력
print("초기 데이터:", initial_data)
print("정렬된 데이터:", sorted_data)
print(f"{end - start:.5f} sec")

# 초기 상태와 최종 상태 비교 (그래프)
plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.bar(range(len(initial_data)), initial_data, color='blue')
plt.title("Initial State")
plt.subplot(1, 2, 2)
plt.bar(range(len(sorted_data)), sorted_data, color='green')
plt.title("Final State")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import time

# 쉘 정렬 함수
def shell_sort(arr):
    n = len(arr)
    gap = n // 2  # 초기 간격 설정

    while gap > 0:
        for i in range(gap, n):
            temp = arr[i]
            j = i
            # gap 간격으로 떨어져 있는 요소들을 삽입 정렬
            while j >= gap and arr[j - gap] > temp:
                arr[j] = arr[j - gap]
                j -= gap
            arr[j] = temp
        gap //= 2  # 간격을 줄임
    return arr

# 사용자에게 데이터 개수 입력 받기
data_size = int(input("정렬할 데이터의 개수를 입력하세요: "))

# 1부터 1000 사이의 랜덤한 정수를 data_size 개 생성
data = np.random.randint(1, 1001, size=data_size)
initial_data = data.copy()

# 쉘 정렬 실행
start = time.time()
sorted_data = shell_sort(data)
end = time.time()

# 초기 상태와 최종 상태를 배열로 출력
print("초기 데이터:", initial_data)
print("정렬된 데이터:", sorted_data)
print(f"{end - start:.5f} sec")

# 초기 상태와 최종 상태 비교 (그래프)
plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.bar(range(len(initial_data)), initial_data, color='blue')
plt.title("Initial State")
plt.subplot(1, 2, 2)
plt.bar(range(len(sorted_data)), sorted_data, color='green')
plt.title("Final State")
plt.show()
```

```python
import tkinter as tk
import time

def heapify(arr, n, i, canvas, node_positions, highlighted=None):
    """Heapify 함수: 최대 힙을 유지"""
    largest = i
    left = 2 * i + 1
    right = 2 * i + 2

    if left < n and arr[left] > arr[largest]:
        largest = left
    if right < n and arr[right] > arr[largest]:
        largest = right

    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]
        draw_heap(arr, canvas, node_positions, highlighted=(i, largest))
        time.sleep(0.8)
        heapify(arr, n, largest, canvas, node_positions, highlighted)

def heap_sort(arr, canvas):
    """Heap Sort 알고리즘"""
    n = len(arr)

    # 힙 빌드
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i, canvas, calculate_positions(len(arr)))

    # 요소를 하나씩 정렬
    for i in range(n - 1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]
        draw_heap(arr, canvas, calculate_positions(len(arr)), highlighted=(0, i))
        time.sleep(0.8)
        heapify(arr, i, 0, canvas, calculate_positions(len(arr)))

def calculate_positions(size):
    """힙 트리를 그리기 위한 노드 위치 계산 (가운데 정렬)"""
    positions = {}
    depth = 0
    level = 1
    index = 0

    while index < size:
        for i in range(level):
            if index >= size:
                break
            x = 350 + (700 // (2 ** (depth + 1))) * (i - (level - 1) / 2)
            y = 50 + 100 * depth
            positions[index] = (x, y)
            index += 1
        level *= 2
        depth += 1
    return positions

def draw_heap(arr, canvas, node_positions, highlighted=None):
    """힙 트리와 배열 상태를 그리기 (변경된 요소 강조)"""
    canvas.delete("all")

    # 트리 노드 그리기
    for i, val in enumerate(arr):
        x, y = node_positions[i]
        color = "red" if highlighted and i in highlighted else "lightblue"
        canvas.create_oval(x-20, y-20, x+20, y+20, fill=color, outline="black")
        canvas.create_text(x, y, text=str(val), font=("Arial", 12))
        if 2 * i + 1 < len(arr):  # 왼쪽 자식
            child_x, child_y = node_positions[2 * i + 1]
            canvas.create_line(x, y, child_x, child_y)
        if 2 * i + 2 < len(arr):  # 오른쪽 자식
            child_x, child_y = node_positions[2 * i + 2]
            canvas.create_line(x, y, child_x, child_y)

    # 배열 상태 그리기
    for i, val in enumerate(arr):
        x1, y1 = 50 + i * 40, 400
        x2, y2 = x1 + 30, 430
        color = "orange" if highlighted and i in highlighted else "lightgray"
        canvas.create_rectangle(x1, y1, x2, y2, fill=color, outline="black")
        canvas.create_text((x1 + x2) / 2, (y1 + y2) / 2, text=str(val), font=("Arial", 10))

    canvas.update()

def start_sorting():
    """정렬 시작"""
    array_input = entry.get()
    try:
        arr = list(map(int, array_input.split()))
        node_positions = calculate_positions(len(arr))

        # 새 창에서 시각화
        visualization_window = tk.Toplevel(root)
        visualization_window.title("Heap Sort Visualization")
        canvas = tk.Canvas(visualization_window, width=700, height=500, bg="white")
        canvas.pack()

        # 힙 정렬 수행 및 시각화
        draw_heap(arr, canvas, node_positions)
        time.sleep(1)
        heap_sort(arr, canvas)

        # 정렬 완료 메시지
        draw_heap(arr, canvas, node_positions)
        canvas.create_text(350, 460, text="정렬 완료!", font=("Arial", 16), fill="green")

    except ValueError:
        result_label.config(text="올바른 숫자 배열을 입력하세요!")

# Tkinter 기본 설정
root = tk.Tk()
root.title("Heap Sort Visualization")

# 사용자 입력 창
frame = tk.Frame(root)
frame.pack(pady=10)

entry_label = tk.Label(frame, text="배열 입력 (숫자 사이 공백):")
entry_label.pack(side=tk.LEFT, padx=5)

entry = tk.Entry(frame, width=30)
entry.pack(side=tk.LEFT, padx=5)

sort_button = tk.Button(frame, text="정렬 시작", command=start_sorting)
sort_button.pack(side=tk.LEFT, padx=5)

result_label = tk.Label(root, text="", fg="red")
result_label.pack(pady=10)

# Tkinter 실행
root.mainloop()
```
