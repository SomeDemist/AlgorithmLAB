```python
import tkinter as tk
from tkinter import messagebox

# 동적 프로그래밍으로 피보나치 수열 생성
def fibonacci_dp(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    elif n == 2:
        return [0, 1]
    
    fib = [0] * n
    fib[0], fib[1] = 0, 1
    
    for i in range(2, n):
        fib[i] = fib[i - 1] + fib[i - 2]
    
    return fib

# 피보나치 수열을 출력하는 함수
def show_fibonacci():
    try:
        n = int(entry.get())
        if n <= 0:
            raise ValueError
        
        # 피보나치 수열 계산
        fib_sequence = fibonacci_dp(n)
        
        # 결과를 새 창에 출력
        result_window = tk.Toplevel(root)
        result_window.title(f"Fibonacci Sequence up to {n} terms")
        
        result_text = tk.Text(result_window, wrap="word", height=10, width=50)
        result_text.pack(padx=10, pady=10)
        
        # 수열을 출력할 텍스트 형식으로 변환
        result_text.insert(tk.END, f"Fibonacci sequence up to {n} terms:\n")
        result_text.insert(tk.END, ', '.join(map(str, fib_sequence)))
        result_text.config(state="disabled")  # 편집 불가능하게 설정
        
    except ValueError:
        messagebox.showerror("입력 오류", "양의 정수를 넣어주세요.")

# GUI 설정
root = tk.Tk()
root.title("피보나치 수열 생성기")

# 입력 필드와 버튼 추가
label = tk.Label(root, text="양의 정수를 입력해주세요 :")
label.pack(pady=5)

entry = tk.Entry(root, width=20)
entry.pack(pady=5)

generate_button = tk.Button(root, text="생성", command=show_fibonacci)
generate_button.pack(pady=10)

root.mainloop()
```
