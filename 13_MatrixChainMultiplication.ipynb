```python
import sys

def matrix_chain_order(p):
    n = len(p) - 1  # 행렬의 개수
    m = [[0] * n for _ in range(n)]  # 연산 횟수 저장
    s = [[0] * n for _ in range(n)]  # 최적 분할 위치 저장
    
    # 부분 문제의 크기가 1일 때는 연산이 필요 없으므로 0으로 초기화되어 있음.
    for l in range(2, n + 1):  # l은 곱셈 체인의 길이
        for i in range(n - l + 1):
            j = i + l - 1
            m[i][j] = sys.maxsize  # 최소값을 찾기 위해 초기화
            
            # 최적 분할 지점을 찾기 위해 k를 탐색
            for k in range(i, j):
                q = m[i][k] + m[k+1][j] + p[i] * p[k+1] * p[j+1]
                if q < m[i][j]:
                    m[i][j] = q
                    s[i][j] = k  # 최적 분할 지점 기록

    return m, s

def print_optimal_order(s, i, j):
    if i == j:
        return f"A{i+1}"
    else:
        return f"({print_optimal_order(s, i, s[i][j])} x {print_optimal_order(s, s[i][j] + 1, j)})"

def main():
    # 사용자로부터 행렬의 개수와 각 행렬의 크기 입력 받기
    n = int(input("행렬의 수를 입력하세요 : "))
    p = []
    print("각 행렬의 크기를 입력하세요 (예: 10 20 = 10 x 20) :")
    for i in range(n):
        dims = list(map(int, input(f"행렬 {i+1}의 크기 : ").split()))
        p.append(dims[0])
        if i == n - 1:
            p.append(dims[1])
    
    # 연속 행렬 곱셈 최적화 계산
    m, s = matrix_chain_order(p)
    
    # 결과 출력
    print(f"연속 행렬 곱셈의 최소 연산 횟수 : {m[0][n-1]}")
    print(f"최적의 곱셈 순서 : {print_optimal_order(s, 0, n-1)}")

if __name__ == "__main__":
    main()
```
