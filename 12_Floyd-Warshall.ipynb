```python
import tkinter as tk
from tkinter import messagebox
import numpy as np

# Floyd-Warshall 알고리즘 함수
def floyd_warshall(matrix, n):
    dist = matrix.copy()  # 초기 거리 행렬
    for k in range(n):
        for i in range(n):
            for j in range(n):
                # dist[i][j]를 경유지 k를 통해 갱신
                dist[i][j] = min(dist[i][j], dist[i][k] + dist[k][j])
    return dist

# 결과 행렬 출력
def display_result(dist):
    # 기존 결과가 있을 경우 제거
    for widget in result_frame.winfo_children():
        widget.destroy()
    
    # 결과 행렬 출력
    for i in range(len(dist)):
        for j in range(len(dist)):
            value = dist[i][j]
            # 자연수일 경우 소수점 없이 출력
            text = f"{int(value)}" if value == int(value) else f"{value:.2f}"
            tk.Label(result_frame, text=text, font=("Arial", 12), width=5).grid(row=i, column=j, padx=5, pady=5)

# 행렬 설정 함수
def set_matrix_size():
    global matrix, n, entries
    
    try:
        n = int(entry_n.get())
        matrix = np.full((n, n), float('inf'))
        
        # 대각선 0으로 설정
        for i in range(n):
            matrix[i][i] = 0
        
        # 기존 입력 필드 제거
        for widget in matrix_frame.winfo_children():
            widget.destroy()
        
        # 행렬 입력 필드 생성
        entries = []
        for i in range(n):
            row_entries = []
            for j in range(n):
                entry = tk.Entry(matrix_frame, width=5, font=("Arial", 12))
                entry.grid(row=i, column=j, padx=5, pady=5)
                row_entries.append(entry)
            entries.append(row_entries)
    
    except ValueError:
        messagebox.showerror("오류", "정수를 입력해 주세요.")

# 값 저장 함수
def save_values():
    global matrix
    for i in range(n):
        for j in range(n):
            try:
                value = float(entries[i][j].get())
                matrix[i][j] = value
            except ValueError:
                matrix[i][j] = float('inf')

# Floyd-Warshall 알고리즘 실행 함수
def run_algorithm():
    if matrix is None:
        messagebox.showerror("오류", "먼저 행렬을 설정해 주세요.")
    else:
        save_values()  # 현재 입력된 값을 행렬에 저장
        result = floyd_warshall(matrix, n)
        display_result(result)

# 프로그램 종료 함수
def exit_program():
    root.destroy()

# GUI 설정
root = tk.Tk()
root.title("Floyd-Warshall 알고리즘")
root.geometry("600x600")  # 창 크기 조정

# 행렬 크기 설정
tk.Label(root, text="행렬 크기 n x n 설정:", font=("Arial", 12)).pack(pady=5)
entry_n = tk.Entry(root, font=("Arial", 12))
entry_n.pack(pady=5)

# 버튼 추가
tk.Button(root, text="크기 설정", command=set_matrix_size, font=("Arial", 12)).pack(pady=5)

# 행렬 입력 프레임
matrix_frame = tk.Frame(root)
matrix_frame.pack(pady=10)

# 실행 버튼 추가
tk.Button(root, text="알고리즘 실행", command=run_algorithm, font=("Arial", 12)).pack(pady=10)

# 결과 출력 프레임
result_frame = tk.Frame(root)
result_frame.pack(pady=10)

# 종료 버튼 추가
tk.Button(root, text="종료", command=exit_program, font=("Arial", 12)).pack(pady=10)

# 메인 루프 실행
root.mainloop()
```
